```python
def forward_chaining(rules, KB):

    inferred_new_fact = True
    known_facts = set(KB)

    print(f"Initial Knowledge Base: {known_facts}\n")

    step = 1
    while inferred_new_fact:
        inferred_new_fact = False
        for rule in rules:
            premises = rule['if']
            conclusion = rule['then']

            if premises.issubset(known_facts) and conclusion not in known_facts:
                known_facts.add(conclusion)
                print(f"Step {step}: Rule [{', '.join(premises)} => {conclusion}] triggered. Added: '{conclusion}'")
                inferred_new_fact = True
                step += 1
                break

    print(f"\nFinal Knowledge Base: {known_facts}")
    return known_facts


rules = [
    {'if': {'A', 'B'}, 'then': 'C'},
    {'if': {'C'}, 'then': 'D'},
    {'if': {'D', 'E'}, 'then': 'F'}
]


initial_facts = {'A', 'B', 'E'}


final_facts = forward_chaining(rules, initial_facts)
```
