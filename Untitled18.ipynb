```python
def resolve(c1, c2):
    for literal in c1:
        if literal.startswith("~"):
            complementary = literal[1:]
        else:
            complementary = "~" + literal

        if complementary in c2:
            new_clause = [
                x for x in c1 if x != literal
            ] + [
                x for x in c2 if x != complementary
            ]
            return new_clause

    return None


def resolution(kb, query):

    clauses = kb + [["~" + query]]

    print("Initial Clauses:")
    for c in clauses:
        print(c)

    while True:
        new_clauses = []

        for i in range(len(clauses)):
            for j in range(i + 1, len(clauses)):

                result = resolve(clauses[i], clauses[j])

                if result is not None:

                    if len(result) == 0:
                        print("\nEmpty clause obtained!")
                        print("Query is PROVED.")
                        return True

                    if result not in clauses and result not in new_clauses:
                        new_clauses.append(result)

        if not new_clauses:
            print("\nQuery cannot be proved.")
            return False

        clauses.extend(new_clauses)
KB = [
    ["HighHumidity", "Cloudy"],
    ["~Cloudy", "Rain"],
    ["~HighHumidity", "Hot"],
    ["~Hot"]
]
query = "Rain"

resolution(KB, query)
```
